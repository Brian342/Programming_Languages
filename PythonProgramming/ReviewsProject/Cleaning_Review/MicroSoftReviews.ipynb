```python
# calculating total null values on the dataset
def null_values(df):
    """This function returns
    total null values on the dataset"""
    print(df.isna().sum())
null_values(MicroRev)
```

```python
# dropping dublicate values
def duplicate_values(df):
    """This function drops all the
    duplicate values on the dataset"""
    df.dropna(inplace=True)
duplicate_values(MicroRev)
# CisRev.dropna(inplace=True)
```

```python
# changing the datatype of the time column
def time_change(df, col="time"):
    df[col] = pd.to_datetime(df[col], format='%b %d, %Y', errors="coerce")
    return df
Rev = time_change(MicroRev)
```

```python
# Rename the headers
MicroRev.rename(columns={"Job Title":"Job Title", "Job Ratings":"Job Rating", "time":"Time", "JobStatus":"Job Status",
                      "Pros":"Pros","Cons":"Cons"},inplace=True)
```

```python
# checking the unique values in the dataset
for job in MicroRev["Job Title"].unique():
    print(job)
```

```python
# capitalizing the dataset
def capital_letter(df, col="Job Title"):
    df[[col]] = df[[col]].apply(lambda c: c.str.title())
    return df
capital_letter(MicroRev)
```

```python
# rename detail
MicroRev["Job Title"] = MicroRev["Job Title"].replace({
    "Tdr Trade Development Representative":"(Tdr) Trade Development Representative",
    "Qaengineer":"Qa Engineer","Çustomer Service Representative":"Customer Service Representative",
    "Sofatware Engineer":"Software Engineer","Softeware Enginee":"Software Engineer",
    "Ftth Engineer":"Ftth (Fiber to the Home) Engineer","Software qa engineer":"Software Qa Engineer",
    "Tdr":"Trade Development Representative","Itspecialist":"IT Specialist",
    "KEY NOT FOUND: jobLine.seasonal-current":"(Job) seasonal_current","I T Analyst":"IT Analyst",
    "It Specialist":"IT Specialist"
    })
```

```python
# SafRev["Pros"] = SafRev["Pros"].str.replace("\r", "", regex=False)
def remove_char(df, col="Pros", col2="Cons"):
    pattern = r"[\r\*\-\(\)]"
    df[["Pros", "Cons"]] = df[["Pros", "Cons"]].apply(
        lambda s: s.str.replace(pattern, "", regex=True))
    return df
remove_char(MicroRev)
```

```python
# SafRev["Pros"] = SafRev["Pros"].str.replace("\r", "", regex=False)
def remove_char(df, col="Pros", col2="Cons"):
    df[[col, col2]] = df[[col, col2]].apply(
        lambda s: s.str.replace("\n", "", regex=False))
    return df
remove_char(MicroRev)
```

```python
hiphen = MicroRev[MicroRev["Pros"].str.startswith("-")]
hiphen
```

```python
# check duplicate
def checkDuplicate(df):
    df.duplicated()
    return df
checkDuplicate(MicroRev)
```

```python
# check duplicate sum
def checkDuplicateSum(df):
    Total = df.duplicated().sum()
    print(f"Sum of Duplicate_Values = {Total}")
checkDuplicateSum(MicroRev)
```

```python
# drop_duplicate
def DropDuplicate(df):
    df.drop_duplicates(subset=["Job Title"], inplace=True)
    return df
DropDuplicate(MicroRev)
```
