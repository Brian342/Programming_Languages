```python
# SafRev["Pros"] = SafRev["Pros"].str.replace("\r", "", regex=False)
def remove_char(df, col="Pros", col2="Cons"):
    pattern = r"[\r\*\-\(\)]"
    df[["Pros", "Cons"]] = df[["Pros", "Cons"]].apply(
        lambda s: s.str.replace(pattern, "", regex=True))
    return df
remove_char(kcbRev)
```

```python
# SafRev["Pros"] = SafRev["Pros"].str.replace("\r", "", regex=False)
def remove_char(df, col="Pros", col2="Cons"):
    df[["Pros", "Cons"]] = df[["Pros", "Cons"]].apply(
        lambda s: s.str.replace("\n", "", regex=True))
    return df
remove_char(kcbRev)
```

```python
# check duplicate sum
def checkDuplicateSum(df: pd.DataFrame) -> int:
    duplicate_count = df.duplicated().sum()
    return 'Sum of Duplicate_Values =', duplicate_count
checkDuplicateSum(kcbRev)
```

```python
def drop_duplicate(df: pd.DataFrame) -> int:
    """This function drops the duplicated 
        values in the dataset
    Args = df for the dataset
    return = this function returns a dataframe 
        withiout the duplicate values"""
    drop_duplicates = df.drop_duplicates(subset = ["Job Title"], inplace=True)
    return drop_duplicates
drop_duplicate(kcbRev)
```
