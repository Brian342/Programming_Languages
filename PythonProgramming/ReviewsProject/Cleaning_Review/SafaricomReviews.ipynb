```python
# Replacing the \n on the Cons column
# SafRev["Pros"]=SafRev["Pros"].str.replace("\n", ",", regex=False)
def replace_str(df, col1="Cons", col2 ="Pros"):
    df[[col1, col2]] = df[[col1, col2]].apply(lambda s: s.str.replace("\n", " ", regex=False))
    return df
replace_str(SafRev)
```

```python
# capitalize the dataset
def capital_letter(df, col="Pros", col2="Cons"):
    df[[col, col2]] = df[[col, col2]].apply(lambda s: s.str.title()) 
    return df
capital_letter(SafRev)
# SafRev['Cons'].str.title()
```

```python
# check duplicate
def checkDuplicate(df):
    df.duplicated()
    return df
checkDuplicate(SafRev)
```

```python
# check duplicate sum
def checkDuplicateSum(df):
    Total = df.duplicated().sum()
    print(f"Sum of Duplicate_Values = {Total}")
checkDuplicateSum(SafRev)
```

```python
# drop_duplicate
def DropDuplicate(df):
    df.drop_duplicates(subset=["Job Title"], inplace=True)
    return df
DropDuplicate(SafRev)
```

```python
# SafRev["Pros"] = SafRev["Pros"].str.replace("\r", "", regex=False)
def remove_char(df, col="Pros", col2="Cons"):
    pattern = r"[\r\*\-\(\)]"
    df[["Pros", "Cons"]] = df[["Pros", "Cons"]].apply(
        lambda s: s.str.replace(pattern, "", regex=True))
    return df
remove_char(SafRev)
```
